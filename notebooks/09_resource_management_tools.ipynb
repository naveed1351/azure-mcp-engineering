This notebook covers the tools you'll reach for constantly regardless of workload: subscriptions, resource groups, RBAC, and Azure best-practices guidance, plus a look at compute/AKS namespaces.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect

async def call(namespace, tool, args=None):
    async with connect(namespaces=[namespace], read_only=True) as client:
        return await client.call_tool(tool, args or {})

print((await call("subscription", "azmcp_subscription_list")).content)
```
