## The three roles: Host, Client, Server

- **Host** — the application the user interacts with (VS Code, or a Python script you write).
- **Client** — lives inside the host; manages one connection to one server.
- **Server** — exposes tools, resources, and prompts (the Azure MCP Server is a server).

In this course, **you** will write the host *and* the client (in Python), and connect to Microsoft's Azure MCP Server.

## Architecture at a glance

```text
+------------------ Host (your Python process) ------------------+
|                                                                  |
|  +------------+   JSON-RPC over stdio   +--------------------+   |
|  | MCP Client | <---------------------->|  Azure MCP Server  |   |
|  +------------+                          +--------------------+   |
|        ^                                                          |
|        | tool schemas / results                                  |
|        v                                                          |
|  +------------+                                                   |
|  |    LLM     |  decides which tool(s) to call                    |
|  +------------+                                                   |
+------------------------------------------------------------------+
```

## Lifecycle of a single tool call

1. Client connects and calls `initialize()`.
2. Client calls `list_tools()` — the server returns names, descriptions, JSON Schemas.
3. The host hands those tool definitions to the LLM alongside the user prompt.
4. If needed, the LLM emits a *tool call* (name + JSON arguments).
5. The client calls `call_tool(name, arguments)`; the server does the real work and returns a result.
6. The result is fed back to the LLM, which produces a final natural-language answer.

## Why MCP for Azure?

Without MCP, an "Azure copilot" needs bespoke integration code for every Azure SDK it wants to expose to an LLM, plus its own auth/retry/error-handling layer. The Azure MCP Server does that work once and exposes a consistent set of tools that *any* MCP client can call the same way — GitHub Copilot agent mode, the OpenAI Agents SDK, Semantic Kernel, or the plain Python client you'll build in notebook 03.
