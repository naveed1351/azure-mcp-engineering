# 08 — Storage and Key Vault Tools

## The `storage` namespace

Tools for Azure Storage accounts, blob containers, blobs, queues, and tables. All examples below use `read_only=True` and never write data.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect

async def list_storage_accounts():
    async with connect(namespaces=["storage"], read_only=True) as client:
        return await client.call_tool("azmcp_storage_account_list", {})

print((await list_storage_accounts()).content)
```

```python
# Once you have an account name, list its blob containers (and then blobs
# within a container). Replace 'account' with a real storage account name.
async def list_containers(account: str):
    async with connect(namespaces=["storage"], read_only=True) as client:
        return await client.call_tool(
            "azmcp_storage_blob_container_list", {"account": account}
        )

# print((await list_containers("mystorageaccount")).content)
print("Uncomment the line above with a real account name to try it.")
```

## The `keyvault` namespace

Key Vault tools manage vaults, secrets, keys, and certificates. **Reading a secret's value triggers elicitation** (a user-confirmation prompt) in interactive clients — see [`docs/05_tools_reference.md`](../docs/05_tools_reference.md#elicitation-user-confirmation-for-sensitive-data).

```python
async def list_key_vaults():
    async with connect(namespaces=["keyvault"], read_only=True) as client:
        return await client.call_tool("azmcp_keyvault_list", {})

print((await list_key_vaults()).content)
```

## RBAC requirements

| Operation | Minimum role |
| --- | --- |
| List storage accounts | Reader |
| List/read blobs | Storage Blob Data Reader |
| List Key Vaults | Reader |
| Read a secret value | Key Vault Secrets User (+ elicitation approval) |
