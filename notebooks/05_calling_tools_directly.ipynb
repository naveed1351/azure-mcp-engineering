# 05 — Calling Tools Directly (No LLM)

Before wiring up an LLM, it's worth calling tools directly with `call_tool(name, arguments)` so you understand exactly what a tool call looks like on the wire.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect

async def list_resource_groups():
    async with connect(namespaces=["group"], read_only=True) as client:
        return await client.call_tool("azmcp_group_list", {})

result = await list_resource_groups()
print(result.content)
```

```python
async def list_storage_accounts():
    async with connect(namespaces=["storage"], read_only=True) as client:
        return await client.call_tool("azmcp_storage_account_list", {})

storage_result = await list_storage_accounts()
print(storage_result.content)
```

## Parsing `CallToolResult.content`

`call_tool` returns a `CallToolResult` whose `.content` is a list of MCP content blocks (usually `TextContent` with a `.text` field containing JSON or plain text). Always treat it as structured data you may need to `json.loads(...)`, not guaranteed-plain-text.

```python
# Defensive error handling: tool calls can fail (auth, RBAC, throttling, typos in
# arguments) and should never crash a whole notebook/agent run.
async def safe_call(namespace, tool_name, arguments):
    try:
        async with connect(namespaces=[namespace], read_only=True) as client:
            return await client.call_tool(tool_name, arguments)
    except Exception as exc:
        print(f"Tool call failed: {exc}")
        return None

await safe_call("storage", "azmcp_storage_account_get", {"account": "does-not-exist"})
```
