# 12 — Server Modes and Namespaces

| Mode | Behavior |
| --- | --- |
| `namespace` (default) | One entry point per namespace; keeps the tool list small. |
| `consolidated` | Merges tools across namespaces into fewer, broader entry points. |
| `all` | Every individual tool is its own top-level MCP tool (most granular). |
| `single` | Exactly one tool, chosen via the `Tool` parameter. |

See [`docs/07_server_modes_and_advanced_config.md`](../docs/07_server_modes_and_advanced_config.md).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import build_server_params

for mode in ("namespace", "consolidated", "all"):
    params = build_server_params(mode=mode)
    print(mode, "->", params.args)
```

```python
# Scoping to only the namespaces a workflow actually needs keeps the tool
# list small and predictable for the model.
scoped = build_server_params(namespaces=["storage", "monitor"], read_only=True)
print(scoped.args)
```

```python
# Single-tool mode: expose exactly one operation. Useful when embedding Azure
# MCP Server capability inside another narrowly-scoped automation.
single_tool_args = [
    "-y", "@azure/mcp@latest", "server", "start",
    "--mode", "single", "--tool", "azmcp_storage_account_get",
]
print(single_tool_args)
```
