# 10 — Data and Analytics Tools

Covers the database and analytics namespaces: `cosmos`, `sql`, `mysql`, `postgres`, `kusto` (Azure Data Explorer), and `monitor` (Log Analytics / KQL).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect

async def call(namespace, tool, args=None):
    async with connect(namespaces=[namespace], read_only=True) as client:
        return await client.call_tool(tool, args or {})

print((await call("cosmos", "azmcp_cosmos_account_list")).content)
```

```python
print((await call("sql", "azmcp_sql_server_list")).content)
```

```python
print((await call("kusto", "azmcp_kusto_cluster_list")).content)
```

```python
# Query Log Analytics for recent exceptions using KQL.
query = "AppExceptions | where TimeGenerated > ago(1h) | take 20"
print((await call("monitor", "azmcp_monitor_log_query", {"query": query})).content)
```

## Summary

| Namespace | Data model |
| --- | --- |
| `cosmos` | Accounts → databases → containers → documents |
| `sql` / `mysql` / `postgres` | Servers → databases → tables |
| `kusto` | Clusters → databases → tables (KQL) |
| `monitor` | Workspaces (KQL over logs/metrics) |
