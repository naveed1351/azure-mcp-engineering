# 04 — Exploring Available Tools

Now that we can connect, let's discover what the server can actually do: list every tool, group them by namespace, and search by keyword.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect
from src.tool_catalog import group_tools_by_namespace, search_tools, export_catalog_to_json

async def list_all_tools():
    async with connect(read_only=True) as client:
        return await client.list_tools()

tools = await list_all_tools()
print(f"Discovered {len(tools)} tools")
```

```python
grouped = group_tools_by_namespace(tools)
for namespace, infos in grouped.items():
    print(f"{namespace:>20}: {len(infos)} tools")
```

```python
# Search across names + descriptions for a keyword.
matches = search_tools(tools, "blob")
for m in matches:
    print(m.name, "-", m.description)
```

```python
# Inspect one tool's full JSON Schema (used later to build OpenAI function
# definitions in notebook 06).
import json

example = tools[0]
print(example.name)
print(json.dumps(example.inputSchema, indent=2))
```

```python
# Save the full catalog for offline reference / diffing between server versions.
out_path = export_catalog_to_json(tools, Path.cwd() / "tool_catalog.json")
print(f"Saved catalog to {out_path}")
```
