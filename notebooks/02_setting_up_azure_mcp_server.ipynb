# 02 — Setting Up the Azure MCP Server

You do **not** need to separately install the Azure MCP Server: every example in this course launches it on demand with `npx -y @azure/mcp@latest server start`. This notebook verifies your toolchain and signs you in.

> See [`docs/03_installation_and_setup.md`](../docs/03_installation_and_setup.md) and [`docs/04_authentication.md`](../docs/04_authentication.md).

```python
import subprocess

def show(cmd: list[str]) -> None:
    try:
        out = subprocess.run(cmd, capture_output=True, text=True, timeout=30)
        print(f"$ {' '.join(cmd)}\n{(out.stdout or out.stderr).strip()}\n")
    except FileNotFoundError:
        print(f"$ {' '.join(cmd)}\nNOT FOUND\n")

show(["node", "--version"])
show(["npx", "--version"])
```

```python
# Confirms you're signed in to Azure and shows the active subscription.
# Run `az login` in a terminal first if this fails.
show(["az", "account", "show", "--output", "table"])
```
