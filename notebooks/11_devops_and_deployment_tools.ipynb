```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.mcp_client import connect

async def call(namespace, tool, args=None):
    async with connect(namespaces=[namespace], read_only=True) as client:
        return await client.call_tool(tool, args or {})

# azd install/usage guidance for deployment scenarios.
print((await call("extension", "azmcp_extension_azd_install_guidance")).content)
```

## A typical CI/CD-adjacent scenario

1. Use `bicepschema`/`azureterraform` tools to validate an IaC template's resource shapes.
2. Use `deploy` (in a controlled, non-read-only environment) to provision resources.
3. Use `monitor`/`applens` (notebook 10 / 09) to confirm the deployment is healthy.
4. Use `loadtesting` to run a load test against the new environment before declaring success.
