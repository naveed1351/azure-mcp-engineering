# 07 — Building a Conversational Agent

Notebook 06 handled exactly one user turn. Real usage needs a growing message history across many turns, referencing earlier context ("now show me their storage accounts" following "list my resource groups"). `src.agent_loop.ConversationAgent` packages notebook 06's logic into a reusable class.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from src.agent_loop import ConversationAgent
from src.mcp_client import AzureMcpClient, build_server_params
```

```python
mcp_client = AzureMcpClient(build_server_params(read_only=True))
await mcp_client.__aenter__()
agent = ConversationAgent(mcp_client)
print("Agent ready.")
```

```python
# Notebooks don't have a real interactive `input()` loop, so we simulate a
# multi-turn conversation with a predefined list of prompts. Each call to
# run_turn() appends to `agent.messages`, so context carries forward.
prompts = [
    "List all of the resource groups in my subscription",
    "Which of those are in the eastus region?",
]
for prompt in prompts:
    print(f"\n> {prompt}")
    answer = await agent.run_turn(prompt)
    print(answer)

await mcp_client.__aexit__(None, None, None)
```
